```python
#1 question
t1=eval(input("Enter a tuple : "))
t2=()

for i in t1:
    if i not in t2:
        t2=t2+(i,)

print("Original Tuple:", t1)
print("New Tuple:", t2)
```

```python
#2 question
students=(("Riya", 85), ("Aman", 92),("Soham", 78), ("Priya", 89))
a=list(students)

for i in range(len(a)):
    for j in range(i + 1, len(a)):
        if a[i][1] < a[j][1]:
            a[i], a[j] = a[j], a[i]

for i in a:
    print(i[0], i[1])
```

```python
#3 question
t=()

for i in range(5):
    n=int(input("Enter a number: "))
    t=t+(n,)

total=sum(t)
average=total / 5

print("Tuple:", t)
print("Sum:", total)
print("Average:", average)
```

```python
#5 question
t=eval(input("Enter a tuple : "))

largest = t[0]
smallest = t[0]

for i in t:
    if i > largest:
        largest = i

    if i < smallest:
        smallest = i

print("Largest element:", largest)
print("Smallest element:", smallest)
```
