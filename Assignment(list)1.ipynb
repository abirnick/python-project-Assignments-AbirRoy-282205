```python
#1 question
l1=[]
for i in range(1,11):
  n=int(input("Enter the number :  "))
  l1.append(n)
max=l1[0]
min=l1[0]
for i in l1:
  if i>max:
    max=i
  if i<min:
    min=i
print("The max element : ",max)
print("The min element : ",min)
```

```python
#2 question
n=int(input("Enter the number of element in list : "))
l1=[]
for i in range(1,n+1):
  n1=int(input("Enter the number : "))
  l1.append(n1)
l2=[]
for i in l1:
  if i not in l2:
    l2.append(i)
  else :
    continue
print(l2)
```

```python
#3 question
n=int(input("Enter the number of element in list : "))
l1=[]
for i in range(1,n+1):
  n1=int(input("Enter the number : "))
  l1.append(n1)
le=[]
l0=[]
for i in l1:
  if i%2==0:
    le.append(i)
  else :
    l0.append(i)
print("The even list : ",le)
print("The odd list : ",l0)
```
