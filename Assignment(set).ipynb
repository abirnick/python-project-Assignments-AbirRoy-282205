```python
#2 question
list1=eval(input("Enter list1:"))
list2=eval(input("Enter list2:"))

A = set(list1)
B = set(list2)

common = A & B

print("Common elements:", common)
```

```python
#3 question
A = {1, 2, 3}
B = {1, 2, 3, 4, 5}

if A.issubset(B):
    print("A is a subset of B")
else:
    print("A is not a subset of B")

if B.issuperset(A):
    print("B is a superset of A")
else:
    print("B is not a superset of A")
```
